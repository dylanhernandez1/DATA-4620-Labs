```python
import numpy as np
import torch
from palmerpenguins import load_penguins
from matplotlib import pyplot as plt

df = load_penguins()

# drop rows with missing values
df.dropna(inplace=True)

# get two features
X = df[['flipper_length_mm','bill_length_mm']].values

# convert species labels to integers
y = df['species'].map({'Adelie':0,'Chinstrap':1,'Gentoo':2}).values

X -= np.mean(X,axis=0)

X = torch.tensor(X).float()
y = torch.tensor(y).long()

MLP_model = torch.nn.Sequential(
    torch.nn.Linear(2,100), # two inputs, 100 outputs
    torch.nn.ReLU(), # activation function
    torch.nn.Linear(100,3) # 100 inputs, 3 outputs
)

linear_model = torch.nn.Sequential(
    torch.nn.Linear(2,3) # two inputs, 3 outputs
)

loss_fn = torch.nn.CrossEntropyLoss()

lr = 1e-2
MLP_opt = torch.optim.SGD(MLP_model.parameters(), lr=lr)
linear_opt = torch.optim.SGD(linear_model.parameters(), lr=lr)

epochs = 100

print("-----------LINEAR MODEL---------------")
for epoch in range(epochs):
    linear_opt.zero_grad() # zero out the gradients

    z = linear_model(X) # compute z values
    loss = loss_fn(z,y) # compute loss

    loss.backward() # compute gradients

    linear_opt.step() # apply gradients

    print(f'epoch {epoch}: loss is {loss.item()}')

print("-----------MLP MODEL---------------")
for epoch in range(epochs):
    MLP_opt.zero_grad() # zero out the gradients

    z = MLP_model(X) # compute z values
    loss = loss_fn(z,y) # compute loss

    loss.backward() # compute gradients

    MLP_opt.step() # apply gradients

    print(f'epoch {epoch}: loss is {loss.item()}')
```

2. Write code to compute the accuracy of each model.


```python
def accuracy(model, X, y):
    model.eval()
    with torch.no_grad():
        z = model(X)
        preds = z.argmax(dim=1)
        return (preds == y).float().mean().item()

print(f'linear model accuracy: {accuracy(linear_model, X, y):.4f}')
```

```python
def accuracy(model, X, y):
    model.eval()
    with torch.no_grad():
        z = model(X)
        preds = z.argmax(dim=1)
        return (preds == y).float().mean().item()

print(f'MLP accuracy: {accuracy(MLP_model, X, y):.4f}')
```
